# Car Price Predictor — Analytics Interface

This notebook demonstrates an analytics interface for predicting the resale
price of used cars using the final trained machine learning model.


The notebook also includes an optional cell to launch the graphical
Streamlit interface locally, if Streamlit is available.


````python
import pandas as pd
import numpy as np
import joblib
import warnings
warnings.filterwarnings(
    "ignore",
    category=UserWarning,
    module="sklearn"
)

````

````python
# ----------- LOAD FINAL MODEL -----------

MODEL_PATH = "bayes_models/HGB_bayes.pkl"
model_pipeline = joblib.load(MODEL_PATH)



# ----------- MANUAL INPUT MODE (Notebook) -----------

input_data = {
    "Brand": "Ford",
    "model": "Focus",
    "car_age": 5,
    "transmission": "Manual",
    "mileage": 45000,
    "fuelType": "Petrol",
    "tax": 150,
    "mpg": 45.3,
    "engineSize": 1.5,
    "previousOwners": 1
}

input_df = pd.DataFrame([input_data])
prediction = model_pipeline.predict(input_df)[0]

print(f"Predicted price: £{prediction:,.0f}")
````
