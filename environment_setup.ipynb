# Environment Setup

notebooks and the optional Streamlit-based user interface.


```python
import sys
import subprocess

print("Installing project dependencies from requirements.txt...")

subprocess.check_call([
    sys.executable,
    "-m",
    "pip",
    "install",
    "-r",
    "requirements.txt"
])

print("All dependencies installed successfully.")
```
